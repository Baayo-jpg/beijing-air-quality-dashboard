## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimanakah tren perubahan bulanan konsentrasi rata-rata polutan PM2.5 di seluruh stasiun pemantauan kualitas udara kota Beijing selama periode Maret 2013 hingga Februari 2017?

## Import Semua Packages/Library yang Digunakan

```python
import glob
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_theme(style='whitegrid')
```

## Data Wrangling

### Gathering Data

#### Load df ...

```python
file_paths = sorted(glob.glob('air-quality-dataset/*.csv') or glob.glob('**/PRSA_Data_*.csv', recursive=True))
dfs = [pd.read_csv(fp) for fp in file_paths]
main_df = pd.concat(dfs, ignore_index=True)

main_df.head()
```

### Assessing Data

#### Identifying ... problem

```python
print("--- Info Data ---")
main_df.info()

print("\n--- Jumlah Missing Values ---")
print(main_df.isna().sum())

print("\n--- Jumlah Duplikat ---")
print(f"Duplikat: {main_df.duplicated().sum()}")

main_df.describe()
```

**Steps to Take:**
- Melakukan imputasi nilai yang hilang (missing values) pada variabel numerik polutan dan cuaca menggunakan metode interpolasi linear (karena data berjenis deret waktu / time series).

- Melakukan imputasi nilai modus pada variabel kategorikal arah angin (`wd`).

- Mengonversi kombinasi kolom `year`, `month`, `day`, dan `hour` menjadi satu kolom berformat Datetime (`datetime`).

- Menghapus kolom `No` yang tidak diperlukan dalam analisis.

**Insight:** (Opsional)

- xxx

### Cleaning Data

#### Fixing ... problem

```python
numeric_cols = [
    'PM2.5',
    'PM10',
    'SO2',
    'NO2',
    'CO',
    'O3',
    'TEMP',
    'PRES',
    'DEWP',
    'RAIN',
    'WSPM',
]
main_df[numeric_cols] = main_df[numeric_cols].interpolate(
    method='linear', limit_direction='both'
)

# 2. Imputasi Kolom Kategorikal (wd) dengan Modus (Menghindari FutureWarning)
main_df['wd'] = main_df['wd'].fillna(main_df['wd'].mode()[0])

# 3. Konversi ke Format Datetime
main_df['datetime'] = pd.to_datetime(main_df[['year', 'month', 'day', 'hour']])

# 4. Drop Kolom 'No'
if 'No' in main_df.columns:
  main_df.drop(columns=['No'], inplace=True)

# Pengecekan Ulang Missing Values
main_df.isna().sum()
```

## Exploratory Data Analysis (EDA)

### Explore ...

```python
station_summary = (
    main_df.groupby('station')[['PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3']]
    .mean()
    .sort_values(by='PM2.5', ascending=False)
    .reset_index()
)
display(station_summary)

# Agregasi Tren Bulanan PM2.5
monthly_df = (
    main_df.groupby(main_df['datetime'].dt.to_period('M'))['PM2.5']
    .mean()
    .reset_index()
)
monthly_df['datetime'] = monthly_df['datetime'].dt.to_timestamp()
monthly_df.head()
```

## Visualization & Explanatory Analysis

### Pertanyaan 1:

```python
plt.figure(figsize=(14, 5))
plt.plot(
    monthly_df['datetime'],
    monthly_df['PM2.5'],
    marker='o',
    linewidth=2,
    color='#2b5c8f',
)
plt.title(
    'Tren Rata-Rata Konsentrasi PM2.5 Bulanan di Beijing (2013 - 2017)',
    fontsize=14,
)
plt.xlabel('Tahun & Bulan', fontsize=11)
plt.ylabel('Konsentrasi PM2.5 (µg/m³)', fontsize=11)
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

### Pertanyaan 2:

```python
plt.figure(figsize=(12, 6))
colors = [
    '#d9534f' if x == station_summary['PM2.5'].max() else '#5bc0de'
    for x in station_summary['PM2.5']
]
sns.barplot(
    x='PM2.5',
    y='station',
    data=station_summary,
    hue='station',
    palette=colors,
    legend=False,
)
plt.title(
    'Rata-Rata Konsentrasi PM2.5 Berdasarkan Stasiun Pemantauan',
    fontsize=14,
)
plt.xlabel('Konsentrasi PM2.5 (µg/m³)', fontsize=11)
plt.ylabel('Stasiun Pemantauan', fontsize=11)
plt.show()
```

## Analisis Lanjutan (Opsional)

```python
def categorize_aqi(pm25):
  if pm25 <= 35.4:
    return 'Baik'
  elif pm25 <= 55.4:
    return 'Sedang'
  elif pm25 <= 150.4:
    return 'Tidak Sehat'
  else:
    return 'Sangat Tidak Sehat'


main_df['AQI_Category'] = main_df['PM2.5'].apply(categorize_aqi)

# Ekspor Data Bersih untuk Dashboard Streamlit
main_df.to_csv('main_data.csv', index=False)

main_df['AQI_Category'].value_counts()
```
